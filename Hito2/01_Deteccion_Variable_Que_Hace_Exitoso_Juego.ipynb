```python
import pandas as pd
import requests
import seaborn as sns
import datetime
import numpy as np

import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
pd.set_option('display.max_columns', 50)
```

Importar datos utiles de juegos de Steam

```python
url = 'https://raw.githubusercontent.com/hectorjimenez12/Mineria_Datos/main/steam.csv'
df = pd.read_csv(url)
df.head()
```

```python
df = df.astype({'english':object,'required_age':object})
print('Info:')
print(df.info())
types_data = df.dtypes.values
```

```python
num_data = df[ [ df.columns[i] for i in range(len(types_data)) if types_data[i] != object  ]  ]
print('Resumen de Atributos Numericos:')
num_data['pos_r/tot_r'] = np.array(num_data['positive_ratings'])/(np.array(num_data['positive_ratings'])+np.array(num_data['negative_ratings']))
numerical_successfull_var = ['por_r/tot_r']
num_data.describe()
```

### Analisis de variable razón de positive rating sobre total ratings:

```python
plt.boxplot(num_data['pos_r/tot_r'])
```

Discretizar data en rangos:

```python
def tag_value(value,ranges,labels):
  for i,val_range in enumerate(ranges):
    if value >= val_range and value <= ranges[i+1]:
      return '[' + str(val_range) + '-' + str(ranges[i+1])+']', labels[i]
print(tag_value(0.9, np.round(np.linspace(0,1,6),1), labels= ['Bajo']*2 + ['Medio']*2 + ['Alto']))
```

```python
ranges, labels = np.round(np.linspace(0,1,6),1), ['Bajo']*2 + ['Medio']*2 + ['Alto']
num_data['posR_interval'] = [tag_value(i,ranges,labels)[0] for i in num_data['pos_r/tot_r'] ]
num_data['posR_label']    = [tag_value(i,ranges,labels)[1] for i in num_data['pos_r/tot_r'] ]
```

Plot de datos

```python
colors = ['b', 'y', 'm', 'c', 'g', 'b', 'r', 'k']
d_color = { k:colors[i] for i,k in enumerate( np.unique(num_data['posR_label'])  ) }
fig, (ax1, ax2, ax3) = plt.subplots(1, 3,figsize=(20,4))
plt.suptitle('Distribucion de razon de comentarios positivos sobre negativos')
num_data['pos_r/tot_r'].plot(kind='density',ax = ax1)
#num_data['posR_cat'].value_counts().sort_values().plot(kind='bar',ax = ax2)
count_labels = num_data.loc[ :, ['posR_interval','posR_label'] ].groupby(by ='posR_interval' ).value_counts()
count_labels.plot(kind='bar',color= [d_color[i[1]] for i in count_labels.index] ,ax = ax2)
num_data['posR_label'].value_counts().sort_values().plot(kind='bar',ax = ax3)
```
